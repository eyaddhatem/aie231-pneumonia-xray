## Model A: small CNN from scratch
`Rescaling(1/255)` -> 3 x [Conv3x3 + ReLU + MaxPool2] with 16, 32, 64 filters -> GlobalAveragePooling -> Dense(64, ReLU) -> Dropout(0.3) -> Dense(1, sigmoid).
**Parameters by hand** (a conv layer has `k*k*in*out + out`):
- Conv1: 3*3*3*16 + 16 = 448
- Conv2: 3*3*16*32 + 32 = 4,640
- Conv3: 3*3*32*64 + 64 = 18,496
- Dense(64): 64*64 + 64 = 4,160
- Output: 64*1 + 1 = 65
- **Total = 27,809** (pooling, dropout and rescaling have no parameters). The cell below asserts it against Keras.
Why this design: convolutions share weights across the image, so 27k parameters see the whole X-ray, versus 262k for the MLP baseline that only saw 64x64 pixels. Global average pooling replaces a huge Flatten+Dense.

```python
def build_cnn():
    inp = keras.Input((IMG_SIZE, IMG_SIZE, 3))
    x = layers.Rescaling(1/255)(inp)
    for f in (16, 32, 64):
        x = layers.Conv2D(f, 3, padding='same', activation='relu')(x)
        x = layers.MaxPooling2D()(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.3)(x)
    return keras.Model(inp, layers.Dense(1, activation='sigmoid')(x), name='cnn_small')
cnn = build_cnn()
assert cnn.count_params() == 27809, cnn.count_params()
cnn.summary()
```

### Training loop with `tf.GradientTape` (spec 3.4)
One step = forward pass, class-weighted binary cross-entropy, gradients of the loss w.r.t. every weight, then Adam updates the weights. We keep the weights from the epoch with the lowest **validation loss** (early stopping, patience 3).

```python
opt = keras.optimizers.Adam(1e-3)
bce = keras.losses.BinaryCrossentropy(reduction='none')

@tf.function
def train_step(x, y):
    with tf.GradientTape() as tape:
        p = cnn(x, training=True)
        w = tf.where(y > 0.5, cw[1], cw[0])
        loss = tf.reduce_mean(bce(y[:, None], p) * w)
    grads = tape.gradient(loss, cnn.trainable_variables)
    opt.apply_gradients(zip(grads, cnn.trainable_variables))
    return loss

def val_loss(model, ds):
    ps = model.predict(ds, verbose=0).ravel()
    return float(keras.losses.BinaryCrossentropy()(yva, ps).numpy()), ps

t0 = time.time(); best, bad, hist = 1e9, 0, []
for ep in range(15):
    ls = [float(train_step(x, y)) for x, y in train_ds]
    vl, ps = val_loss(cnn, val_ds)
    hist.append((np.mean(ls), vl)); print(f'epoch {ep+1}: train loss {np.mean(ls):.4f} | val loss {vl:.4f}')
    if vl < best: best, bad = vl, 0; cnn.save_weights('models/cnn_small.weights.h5')
    else:
        bad += 1
        if bad >= 3: break
cnn.load_weights('models/cnn_small.weights.h5')
t_cnn = time.time() - t0
vl, p_cnn = val_loss(cnn, val_ds)
m_cnn = binary_metrics(yva, p_cnn); print({k: round(v, 4) for k, v in m_cnn.items()})
log_run(LOG, AUTHOR, 'cnn_small', {'filters': [16, 32, 64], 'dropout': 0.3, 'lr': 1e-3, 'batch': 32, 'epochs': len(hist), 'aug': 'rot+zoom+contrast'}, SEED, t_cnn, cnn.count_params(), 'val', m_cnn)
plt.plot(*zip(*hist)); plt.legend(['train', 'val']); plt.title('CNN loss per epoch'); plt.show()
```

## Models B and C: pretrained VGG16 (only VGG16 / ResNet-50 are allowed by the spec)
VGG16 was trained on ImageNet (natural photos). Its early layers detect edges and textures that also help on X-rays. `preprocess_input` converts pixels the way VGG16 expects.
**Parameters by hand:** VGG16 conv base = 14,714,688. New head = GlobalAveragePooling -> Dropout(0.3) -> Dense(1): 512*1 + 1 = **513**.
- **B (frozen):** only the head trains = 513 trainable, 14,715,201 total.
- **C (fine-tune last block):** `block5` has 3 convs, each 3*3*512*512 + 512 = 2,359,808, so 3 x 2,359,808 = 7,079,424. Trainable = 7,079,424 + 513 = **7,079,937**, trained with a tiny learning rate (1e-5) so we nudge, not destroy, the pretrained filters.

```python
def build_vgg():
    base = keras.applications.VGG16(include_top=False, weights='imagenet', input_shape=(IMG_SIZE, IMG_SIZE, 3))
    base.trainable = False
    inp = keras.Input((IMG_SIZE, IMG_SIZE, 3))
    x = layers.Lambda(keras.applications.vgg16.preprocess_input)(inp)
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    return keras.Model(inp, layers.Dense(1, activation='sigmoid')(x), name='vgg16'), base

def fit_eval(model, name, lr, epochs, hp, notes=''):
    model.compile(keras.optimizers.Adam(lr), 'binary_crossentropy', metrics=['accuracy'])
    t0 = time.time()
    h = model.fit(train_ds, validation_data=val_ds, epochs=epochs, class_weight=cw, verbose=2,
                  callbacks=[keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True, monitor='val_loss')])
    secs = time.time() - t0
    p = model.predict(val_ds, verbose=0).ravel(); m = binary_metrics(yva, p)
    trainable = int(sum(np.prod(v.shape) for v in model.trainable_variables))
    print(name, 'trainable params', trainable, {k: round(v, 4) for k, v in m.items()})
    log_run(LOG, AUTHOR, name, dict(hp, lr=lr, batch=32, epochs=len(h.history['loss'])), SEED, secs, model.count_params(), 'val', m, notes=f'trainable={trainable} {notes}')
    return p, m, h, trainable

vgg, base = build_vgg()
p_b, m_b, h_b, tr_b = fit_eval(vgg, 'vgg16_frozen', 1e-3, 8, {'frozen': True})
assert tr_b == 513 and vgg.count_params() == 14715201
vgg.save_weights('models/vgg16_frozen.weights.h5')
```

```python
# Model C continues from the trained frozen model: unfreeze only block5, then train with a tiny lr.
base.trainable = True
for l in base.layers: l.trainable = l.name.startswith('block5')
p_c, m_c, h_c, tr_c = fit_eval(vgg, 'vgg16_finetune_block5', 1e-5, 8, {'unfrozen': 'block5'}, 'continued from vgg16_frozen')
assert tr_c == 7079937, tr_c
vgg.save_weights('models/vgg16_finetune_block5.weights.h5')
```

## Comparison on validation
Primary metric = **recall on pneumonia** (a missed pneumonia case is the costly error), then recall on normal, F1 and AUC. Baselines come from notebook 03 (same validation set).

```python
log = pd.read_csv(LOG)
cols = ['model', 'params', 'accuracy', 'recall_pneumonia', 'recall_normal', 'f1_pneumonia', 'macro_f1', 'auc', 'train_time_s']
display(log[log.split == 'val'][cols].round(3))
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
for p, n in [(p_cnn, 'CNN small'), (p_b, 'VGG16 frozen'), (p_c, 'VGG16 fine-tuned')]:
    RocCurveDisplay.from_predictions(yva, p, name=n, ax=ax[0])
ax[0].set_title('ROC (val)')
for k, (p, n) in enumerate([(p_cnn, 'CNN'), (p_b, 'frozen'), (p_c, 'finetune')]):
    m = binary_metrics(yva, p); ax[1].bar(n, m['recall_pneumonia'], color='C%d' % k); ax[1].text(n, m['recall_pneumonia'] + .005, f"{m['recall_pneumonia']:.3f}", ha='center')
ax[1].set_ylim(0.8, 1.02); ax[1].set_title('Recall on pneumonia (val)'); plt.show()
```

**What to say in the defense:** pretrained VGG16 starts from useful features, so even frozen it needs few epochs; fine-tuning block5 adapts the high-level filters to X-rays, usually gaining a little recall at the cost of more training time and overfitting risk. Whichever model wins here is *not* final: tuning (M5) and the 3-seed test evaluation (M7) decide.
