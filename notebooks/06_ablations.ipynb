# 06 - Ablation study
**Milestone 6 (spec 3.6)** | Owner: Experiment and Tuning Engineer. An ablation removes **one** ingredient from the full recipe and measures what we lose. If removing it changes nothing, it was not earning its place.
Full recipe (from M5): VGG16 pretrained on ImageNet, 3 epochs head-only at lr 1e-3, then fine-tune `block5` at lr 1e-5, dropout 0.5, class weights, augmentation (rotation, zoom, contrast).
Ablations (one change each): **no augmentation**, **no class weights**, **no dropout**, **no fine-tuning** (head only), **no pretraining** (random weights, whole network trained at lr 1e-4).
M5 showed that one seed is noisy, so every setting is run with **2 seeds (42 and 7)** on the same split and we report both. Validation set only; the test set stays untouched. Threshold 0.5 for all rows (the tuned threshold is applied in the final evaluation).
**Runtime: T4 GPU.**

```python
ROOT = find_data_root(download())
inv = build_inventory(ROOT)
df = make_splits(drop_exact_duplicates(inv)); check_no_leakage(df)   # same seed-42 split as every notebook
tr, va = df[df.split == 'train'], df[df.split == 'val']
CW = class_weights(tr); yva = va.label.values
raw_tr = make_dataset(tr, 32, training=True).unbatch().cache().shuffle(len(tr), seed=42, reshuffle_each_iteration=True).batch(32)   # cache single images, re-shuffle every epoch
aug = keras.Sequential([layers.RandomRotation(0.04, seed=42), layers.RandomZoom(0.1, seed=42), layers.RandomContrast(0.15, seed=42)])
DS_AUG = raw_tr.map(lambda x, y: (aug(x, training=True), y), num_parallel_calls=tf.data.AUTOTUNE).prefetch(tf.data.AUTOTUNE)
DS_PLAIN = raw_tr.prefetch(tf.data.AUTOTUNE)
val_ds = make_dataset(va, 32).cache()
print(len(tr), 'train |', len(va), 'val | class weights', CW)
```

```python
def build_vgg(dropout, pretrained=True):
    base = keras.applications.VGG16(include_top=False, weights='imagenet' if pretrained else None, input_shape=(IMG_SIZE, IMG_SIZE, 3))
    base.trainable = not pretrained
    inp = keras.Input((IMG_SIZE, IMG_SIZE, 3))
    x = layers.Lambda(keras.applications.vgg16.preprocess_input)(inp)
    x = base(x, training=not pretrained)
    x = layers.GlobalAveragePooling2D()(x)
    if dropout > 0: x = layers.Dropout(dropout)(x)
    return keras.Model(inp, layers.Dense(1, activation='sigmoid')(x)), base

FULL = dict(augment=True, cw=True, dropout=0.5, finetune=True, pretrained=True)
ABL = {'full': {}, 'no_augmentation': dict(augment=False), 'no_class_weights': dict(cw=False),
       'no_dropout': dict(dropout=0.0), 'no_finetuning': dict(finetune=False), 'no_pretraining': dict(pretrained=False)}
ROWS = []
def run(name, seed):
    cfg = dict(FULL, **ABL[name]); tf.keras.utils.set_random_seed(seed)
    ds = DS_AUG if cfg['augment'] else DS_PLAIN; cw = CW if cfg['cw'] else None; t0 = time.time()
    model, base = build_vgg(cfg['dropout'], cfg['pretrained'])
    es = keras.callbacks.EarlyStopping(patience=2, restore_best_weights=True, monitor='val_loss')
    if not cfg['pretrained']:
        model.compile(keras.optimizers.Adam(1e-4), 'binary_crossentropy')
        h = model.fit(ds, validation_data=val_ds, epochs=10, class_weight=cw, verbose=0, callbacks=[es])
    else:
        model.compile(keras.optimizers.Adam(1e-3), 'binary_crossentropy')
        model.fit(ds, validation_data=val_ds, epochs=3, class_weight=cw, verbose=0)
        if cfg['finetune']:
            for l in base.layers: l.trainable = l.name.startswith('block5')
            base.trainable = True
            model.compile(keras.optimizers.Adam(1e-5), 'binary_crossentropy')
            h = model.fit(ds, validation_data=val_ds, epochs=8, class_weight=cw, verbose=0, callbacks=[es])
    secs = time.time() - t0
    p = model.predict(val_ds, verbose=0).ravel(); m = binary_metrics(yva, p)
    log_run(LOG, AUTHOR, 'ablation_' + name, dict(cfg, seed=seed), seed, secs, model.count_params(), 'val', m)
    ROWS.append(dict(ablation=name, seed=seed, accuracy=m['accuracy'], recall_pneumonia=m['recall_pneumonia'], recall_normal=m['recall_normal'], macro_f1=m['macro_f1'], auc=m['auc'], fn=m['fn'], fp=m['fp'], secs=round(secs)))
    print(name, 'seed', seed, {k: round(float(v), 4) for k, v in m.items() if k in ('accuracy', 'recall_pneumonia', 'recall_normal', 'macro_f1', 'auc', 'fn', 'fp')}, round(secs), 's')
    keras.backend.clear_session()

for seed in (42, 7):
    for name in ABL:
        run(name, seed)
```

## Summary: mean over the two seeds (min to max in the per-run lines above)
Read the **drop versus `full`** per ingredient. A drop smaller than the gap between the two seeds of the same setting is noise.

```python
res = pd.DataFrame(ROWS)
summ = res.groupby('ablation', sort=False).agg(recall_pneumonia=('recall_pneumonia', 'mean'), recall_normal=('recall_normal', 'mean'), macro_f1=('macro_f1', 'mean'), auc=('auc', 'mean'), fn=('fn', 'mean'), fp=('fp', 'mean'),
    seed_gap_recall=('recall_pneumonia', lambda s: s.max() - s.min()), seed_gap_macro_f1=('macro_f1', lambda s: s.max() - s.min()))
summ['d_recall_vs_full'] = summ.recall_pneumonia - summ.loc['full', 'recall_pneumonia']
summ['d_macro_f1_vs_full'] = summ.macro_f1 - summ.loc['full', 'macro_f1']
display(summ.round(4))
print(summ.round(4).to_csv())
ax = summ[['recall_pneumonia', 'recall_normal', 'macro_f1']].plot.bar(figsize=(9, 4)); ax.set_ylim(0.5, 1.0); ax.set_title('Ablations (val, mean of 2 seeds)'); plt.xticks(rotation=30, ha='right'); plt.show()
```
